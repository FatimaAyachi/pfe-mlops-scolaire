```python
import pandas as pd 
import os
os.chdir(r"D:\pfe-mlops-scolaire")
```

```python
# Chargement des fichiers
y_train = pd.read_csv('data/y_train.csv')
y_test = pd.read_csv('data/y_test.csv')
```

```python
# Seuil de réussite (10/20)
seuil = 10
```

```python
# Transformation : 
#   ---> 1 si G3 >= 10
#   ---> 2 si G3 < 10
y_train_class = (y_train['G3'] >= seuil).astype('int')
y_test_class =(y_test['G3'] >= seuil).astype('int')
```

```python
y_train_class.to_csv('data/y_train_binary.csv', index=False)
y_test_class.to_csv('data/y_test_binary_1.csv', index=False)
```
