```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sea
```

```python
#chargement avec le bon separateur
data = pd.read_csv('data/student-mat-before.csv', sep=';')

#Afficher les 5 premieres ligne
print(data.head())

#verificatin des types de col et des val manquantes
print(data.info())
#les val manquante
data.isnull().sum()
```

```python
print(data['G3'].describe())
```

```python
#select only les colonnes numerique
num_cl_of_data = data.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(8, 6))
sea.heatmap(num_cl_of_data.corr(), annot=True, cmap='RdYlGn', fmt='.2f')
plt.title('Matrice de correlation des variables numeruiques')
plt.show()
```
